```python
import os
import glob
import MDAnalysis as mda

base_dir = "/deac/phy/classes/phy320-fa-2026/andrtj24/Project-1-PHY320620-Fall2026/6od3-1"
if not os.path.exists(base_dir):
    base_dir = "/deac/phy/classes/phy320-fa-2026/andrtj24/Project-1-PHY320620-Fall2026/6o3-1"

# Modified to only include '1'
replica_dirs = ['1'] 
universes = {}

for rep in replica_dirs:
    target_dir = os.path.join(base_dir, rep, "namd")
    psf_path = glob.glob(os.path.join(target_dir, "*.psf"))
    dcd_path = sorted(glob.glob(os.path.join(target_dir, "*.dcd")))
    
    if psf_path and dcd_path:
        universes[rep] = mda.Universe(psf_path[0], dcd_path)
        print(f"Loaded Replica {rep}: {len(universes[rep].trajectory)} frames.")

if not universes:
    print("Error: Could not load the trajectory.")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
from MDAnalysis.analysis import distances

# --- Constants ---
KE_KCAL = 332.0637   
DIELECTRIC = 4.0     
CUTOFF = 12.0        
TEMPERATURE = 310.15 
R_GAS = 0.0019872    
RT = R_GAS * TEMPERATURE
T_DELTA_S = 22.0     

all_energy_data = []
print("Calculating Interaction Energies and Thermodynamics for Replica 1...")

for rep, u in universes.items():
    prot_atoms = u.select_atoms("protein")
    dna_atoms = u.select_atoms("nucleic")
    
    q_prot = prot_atoms.charges
    q_dna = dna_atoms.charges
    
    for ts in u.trajectory:
        # Override time calculation to account for the NAMD dcdfreq=5000 (10 ps/frame)
        time_ns = ts.frame * 10.0 / 1000.0
        
        # Calculate pairwise distance matrix
        dist_matrix = distances.distance_array(prot_atoms.positions, dna_atoms.positions)
        
        # Lower bound of 2.5 A prevents vdW explosions from minor steric clashes
        in_range = (dist_matrix > 2.5) & (dist_matrix < CUTOFF)
        if not np.any(in_range):
            continue
            
        i_idx, j_idx = np.where(in_range)
        r_ij = dist_matrix[i_idx, j_idx]
        
        # Electrostatics
        q_ij = q_prot[i_idx] * q_dna[j_idx]
        e_elec_pairs = (KE_KCAL * q_ij) / (DIELECTRIC * (r_ij ** 2))
        e_att_elec = np.sum(e_elec_pairs[q_ij < 0])
        e_rep_elec = np.sum(e_elec_pairs[q_ij > 0])
        
        # Simplified Lennard-Jones (vdW)
        sigma, epsilon = 3.5, 0.15
        sr_term = (sigma / r_ij) ** 6
        e_vdw_pairs = 4 * epsilon * (sr_term**2 - sr_term)
        e_att_vdw = np.sum(e_vdw_pairs[e_vdw_pairs < 0])
        e_rep_vdw = np.sum(e_vdw_pairs[e_vdw_pairs > 0])
        
        net_enthalpy = e_att_elec + e_rep_elec + e_att_vdw + e_rep_vdw
        delta_g = net_enthalpy + T_DELTA_S
        kd_m = np.exp(delta_g / RT)
        
        all_energy_data.append({
            'Time (ns)': time_ns,
            'Elec_Attractive': e_att_elec,
            'Elec_Repulsive': e_rep_elec,
            'Net_Enthalpy_dH': net_enthalpy,
            'Delta_G': delta_g,
            'Kd_M': kd_m
        })

energy_df = pd.DataFrame(all_energy_data)

# --- Summary Statistics ---
mean_dH = energy_df['Net_Enthalpy_dH'].mean()
mean_dG = energy_df['Delta_G'].mean()
mean_Kd_nM = np.exp(mean_dG / RT) * 1e9

summary_df = pd.DataFrame([{
    'Mean ΔH (kcal/mol)': round(mean_dH, 2),
    'Mean ΔG (kcal/mol)': round(mean_dG, 2),
    'Estimated Kd (nM)': round(mean_Kd_nM, 2)
}])

print("\nThermodynamic Summary Table (Replica 1):")
display(summary_df)

# --- Plotting ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Plot 1: Interaction Energies
sns.lineplot(data=energy_df, x='Time (ns)', y='Elec_Attractive', color="navy", label="Attractive (< 0)", ax=ax1)
sns.lineplot(data=energy_df, x='Time (ns)', y='Elec_Repulsive', color="crimson", label="Repulsive (> 0)", ax=ax1)
sns.lineplot(data=energy_df, x='Time (ns)', y='Net_Enthalpy_dH', color="black", label="Net Enthalpy ($\Delta H$)", ax=ax1)
ax1.set_ylabel("Energy (kcal/mol)")
ax1.set_title("Protein-DNA Electrostatic Interactions Over Time")
ax1.grid(True, linestyle="--", alpha=0.5)

# Plot 2: Dissociation Constant (Kd)
sns.lineplot(data=energy_df, x='Time (ns)', y='Kd_M', color="purple", ax=ax2)
ax2.set_yscale('log')
ax2.set_ylabel('$K_d$ (Molar) [Log Scale]')
ax2.set_xlabel('Time (ns)')
ax2.set_title('Estimated Dissociation Constant ($K_d$) Trajectory')
ax2.grid(True, which="both", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("thermodynamics_analysis_rep1.png", dpi=300, bbox_inches='tight')
plt.show()
```
